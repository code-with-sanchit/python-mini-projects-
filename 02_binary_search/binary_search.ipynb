```python
import random 

start = random.randint(0,10)
numbers = list(range(start,101,2))
print("List:",numbers)
```

```python
# Binary Search Function 
def binary_search(arr,x):
    low = 0 
    high = len(arr) - 1

    while low <= high: 
        mid = (low + high) // 2 

        if arr[mid] == x: 
            return mid 
        elif x > arr[mid]: 
            low = mid + 1 
        else: 
            high = mid - 1 
            
    return -1 
```

```python
x = int(input("Enter a number to search:"))
result = binary_search(numbers,x)

if result != -1: 
    print(f"{x} found at index {result}")
else: 
    print(f"{x} is not in the list")
```
