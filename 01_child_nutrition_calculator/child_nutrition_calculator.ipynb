```python
name = input("Enter Child's name")
age = int(input("Enter age in years (0-8):"))
gender = input("Enter gender (M/F):")
height = float(input("Enter height in inches:"))
weight = float(input("Enter weight in pounds:"))

child = {"name":name , "age": age, "gender": gender, "height": height, "weight": weight}
print(child)
```

```python
# BMI (weight in pounds, height in inches)
bmi = weight / (height ** 2) * 703 
print(f"BMI of {name} is {bmi: .2f}")
```

```python
if age < 2: 
    required_calories = 800
elif age < 4: 
    required_calories = 1400 
elif age <= 8: 
    required_calories = 1800 
else: 
    required_calories = None 

if required_calories is None: 
    print("Age is outside the range given in the assignment (0-8).")
else: 
    print(f"Minimum daily calories needed: {required_calories}")
```

```python
calories_per_100g = {
    "milk": 100, 
    "egg": 155, 
    "rice": 130, 
    "lentils": 113,
    "vegetable":85,
    "meat": 143,
}

total_calories = 0 
for food in calories_per_100g: 
    grams = float(input(f"How many grams of {food} did {name} eat? (0 if none):"))
    total_calories += grams / 100 * calories_per_100g[food]

print(f"Total calories eaten today: {total_calories: .2f}")
```

```python
if required_calories is not None: 
    if total_calories < required_calories: 
        shortage = required_calories - total_calories 
        print(f"{name} is UNDERNOURISHED. Short by {shortage:.2f} calories.")
    else:
        print(f"{name} is getting enough nutrition today.")
```
