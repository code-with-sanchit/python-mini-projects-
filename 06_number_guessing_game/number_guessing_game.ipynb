```python
import random
import math

def get_valid_number(message):
    """Keep asking until the user types a real integer."""
    while True:
        try:
            return int(input(message))
        except ValueError:
            print("Please enter a whole number.")

def compare(guess, secret):
    """Return 'high', 'low' or 'correct'."""
    if guess > secret:
        return "high"
    elif guess < secret:
        return "low"
    return "correct"

def difference(guess, secret):
    """How far the guess is from the secret number."""
    return abs(guess - secret)
```

```python
low = get_valid_number("Enter the start of the range (A): ")
high = get_valid_number("Enter the end of the range (B): ")

if low >= high:
    print("A must be smaller than B.")
else:
    secret = random.randint(low, high)

    # Minimum guesses needed with a smart (binary search) strategy
    min_guesses = math.ceil(math.log2(high - low + 1))
    print(f"Guess the number between {low} and {high}. Try to do it in {min_guesses} guesses!")

    score = 100
    guesses = 0

    while True:
        guess = get_valid_number("Your guess: ")
        guesses += 1
        result = compare(guess, secret)

        if result == "correct":
            print(f"You got it in {guesses} guesses! Final score: {score}")
            if guesses <= min_guesses:
                print("Congratulations!")
            else:
                print("Better Luck Next Time!")
            break

        # Wrong guess: reduce score and give a hint
        score = max(score - 10, 0)

        if result == "high":
            print("Try Again! You guessed too high")
        else:
            print("Try Again! You guessed too small")

        # Extra hints
        if difference(guess, secret) <= 5:
            print("Hint: You are very close!")
        if guesses == 2:
            if secret % 2 == 0:
                print("Hint: The number is even.")
            else:
                print("Hint: The number is odd.")
        if guesses == 3:
            print(f"Hint: The number is {'' if secret % 3 == 0 else 'not '}a multiple of 3.")
```
